## **Comprendiendo el negocio**

**Objetivo del negocio**

Una empresa dedicada a la distribución y comercialización de videojuegos desea comprender cuáles son los factores que influyen en el éxito comercial de un videojuego. La organización busca identificar tendencias relacionadas con plataformas, géneros, años de lanzamiento y regiones geográficas para apoyar la toma de decisiones sobre futuras inversiones, campañas de marketing y selección de títulos para su catálogo.

**Objetivos analíticos**

Se busca obtener datos relevantes dentro de la industria, principalmente para responder las siguientes incognitas:

lgunas preguntas que responderá el proyecto:

- ¿Qué plataformas han generado mayores ventas?
- ¿Qué géneros son los más exitosos?
- ¿Cómo evolucionan las ventas a través de los años?
- ¿Qué editoras publican los videojuegos con mayores ventas?
- ¿Existen diferencias entre Norteamérica, Europa y Japón?
- ¿Qué características poseen los videojuegos más vendidos?


# **Comprendiendo los datos**

```python
list(df.columns)
```

Las columnas dentro del dataset y sus significado son las siguientes:

- **Rank**: Clasificación de ventas totales.
- **Name**: El nombre del juego.
- **Platform**: Plataforma de lanzamiento de los juegos.
- **Year**: Año en que se lanzo el juego.
- **Genere**: Genero del juego.
- **Publisher**: Editor del juego.
- **NA_Sales**: Ventas en norte america (en millones).
- **EU_Sales**: Ventas en europa (en millones).
- **JP_Sales**: Ventas en japon.
- **Other_sales**: Ventas en el resto del mundo.
- **Global Sales**: Total de ventas mundiales.
